```python
import pandas as pd
crime = pd.read_csv("crime.csv")
crime.columns
```

```python
# List of columns you actually want to keep
keep_cols = ['Year', 'County', 'Violent_sum', 'Property_sum', 'Homicide_sum', 'LTtotal_sum']

# Filter the dataframe to only these columns
crime = crime[keep_cols]

crime = crime.rename(columns={
    'Violent_sum': 'Total_Violent',
    'Property_sum': 'Total_Property',
    'LTtotal_sum': 'Larceny_Theft',
    'ForRape_sum': 'Rape'})

crime = crime[crime['Year'] > 2009].reset_index()
crime
```

```python
import pandas as pd

# Load the datasets
crime_df = pd.read_csv("/mnt/data/crime.csv")
unemployment_df = pd.read_csv("/mnt/data/Unemployment Data.csv")

# Optional: Standardize column names (lowercase + strip spaces)
crime_df.columns = crime_df.columns.str.strip().str.lower()
unemployment_df.columns = unemployment_df.columns.str.strip().str.lower()

# Optional: Clean merge key columns (remove extra spaces, standardize case)
for col in ['year', 'county', 'month']:
    crime_df[col] = crime_df[col].astype(str).str.strip().str.lower()
    unemployment_df[col] = unemployment_df[col].astype(str).str.strip().str.lower()

# Merge the datasets
merged_df = pd.merge(
    crime_df,
    unemployment_df,
    on=['year', 'county', 'month'],
    how='inner'   # change to 'left' if you want to keep all crime records
)

# Save the merged file
merged_df.to_csv("/mnt/data/merged_output.csv", index=False)

print("Merge complete.")
print("Merged dataset shape:", merged_df.shape)
print(merged_df.head())
```
